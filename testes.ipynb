# Atividade de Python

 Par ou ímpar

```python
numero = int(input("insira um número:"))
if numero % 2 == 0:
    print("esse número é par")
else:
    print("esse número é impar")
```

Situação do aluno

```python

nota1 = float(input("Insira a primeira nota: "))
nota2 = float(input("Insira a segunda nota: "))
media = (nota1 + nota2) / 2
if media >= 7:
    print ("Aprovado")
elif media < 7 and media >= 5:
    print ("Recuperação")
else:
    print ("Reprovado")
```

Maior entre três números

```python

num1 = int(input("Primeiro número: "))
num2 = int(input("Segundo número: "))
num3 = int(input("Terceiro número: "))
maior = num1
if num2 > maior:
    maior = num2
if num3 > maior:
    maior = num3
print("Maior número:", maior)
```

Contagem crescente

```python

N = int(input("Digite um número inteiro positivo: "))
for i in range(1, N + 1):
    print(i)
```

Somando de 1 até N

```python

N = int(input("Digite N: "))
soma = 0
for i in range(1, N + 1):
  soma = soma + i
print("Soma:", soma)
```

Contando números pares

```python

pares = 0
for i in range(10):
    numero = int(input("Digite um número: "))
    if numero % 2 == 0:
        pares = pares + 1
print("Foram digitados", pares, "números pares.")
```

Soma apenas dos positivos

```python

soma = 0
for i in range(8):
    numero = int(input("Digite um número: "))
    if numero > 0:
        soma = soma + numero
print("Soma dos positivos =", soma)
```

Maior valor digitado

```python

maior = None
for i in range(6):
    numero = int(input("Digite um número: "))
    if maior is None or numero > maior:
        maior = numero
print("Maior valor:", maior)
```

Tabuada

```python

numero = int(input("Digite um número: "))
for i in range(1, 11):
    resultado = numero * i
    print(numero, "x", i, "=", resultado)
```

Número secreto

```python

numero_secreto = 7
tentativas = 0
numero = int(input("Tente descobrir o número: "))
while numero != numero_secreto:
    tentativas = tentativas + 1
    if numero < numero_secreto:
        print("Tente um número maior")
    else:
        print("Tente um número menor")
    numero = int(input("Tente novamente: "))
tentativas = tentativas + 1
print("Parabéns! Você acertou.")
print("Tentativas:", tentativas)
```
